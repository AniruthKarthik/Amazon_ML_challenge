## Turnkey Google Colab T4 GPU Execution

**Hardware Requirement:** Ensure GPU runtime is active: `Runtime` -> `Change runtime type` -> select **T4 GPU**.

```python
# Step 1: Verify NVIDIA T4 GPU and Unpack Archive
import os, subprocess, sys

subprocess.run(["nvidia-smi"], check=False)

# Unzip aml_collab.zip
subprocess.run(["unzip", "-q", "-o", "/content/aml_collab.zip", "-d", "/content/aml"], check=True)
os.chdir("/content/aml")

# Install dependencies
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "chimera_submission/code/business_entity_resolution/requirements.txt"], check=True)
print("Environment and data ready!")
```

```python
# Step 2: Run End-to-End Entity Resolution Pipeline with T4 GPU Acceleration
import subprocess, sys

cmd = [
    sys.executable, "chimera_submission/code/business_entity_resolution/src/main.py",
    "--train-dir", "dataset/student_resource/dataset/train",
    "--test-dir", "dataset/student_resource/dataset/test",
    "--output-dir", "chimera_submission/output",
    "--k-folds", "5",
    "--seed", "42",
    "--device", "auto",
    "--n-jobs", "-1",
    "--max-train-queries", "40000"
]
subprocess.run(cmd, check=True)
```

```python
# Step 3: Run Official Submission Validator & Package Output
import subprocess, sys

val_cmd = [
    sys.executable, "utils/validate_submission.py",
    "--matching", "chimera_submission/output/matching_results.tsv",
    "--candidate", "chimera_submission/output/candidate_pairs.tsv",
    "--test-dir", "dataset/student_resource/dataset/test"
]
subprocess.run(val_cmd, check=True)

# Create clean submission archive
subprocess.run([
    "zip", "-q", "-j", "/content/submission_results.zip",
    "chimera_submission/output/matching_results.tsv",
    "chimera_submission/output/candidate_pairs.tsv"
], check=True)

try:
    from google.colab import files
    files.download('/content/submission_results.zip')
except ImportError:
    print("Archive created at /content/submission_results.zip")
```
